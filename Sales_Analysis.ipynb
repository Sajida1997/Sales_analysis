#  Data Cleaning

```python
data.columns
```

```python
data.drop(['Status', 'unnamed1'], axis= 1, inplace= True)
```

```python
data.columns
```

```python
data["Marital_Status"]= data["Marital_Status"].astype("str").str.replace("0","Single")

data["Marital_Status"]= data["Marital_Status"].astype("str").str.replace("1","Married")
```

```python
data
```

```python
data.rename(columns={"Occupation":"Sector"},inplace=True)

data.columns
```

# Data Visualization

```python
plt.figure(figsize=(15,6))

A1 = sns.countplot(x="Zone",data=data,hue="Gender")

for i in A1.containers:
    A1.bar_label(i)

plt.title("Gender vs Zone Wise Count")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(15,6))

A2 = sns.countplot(x='Age Group',data=data,hue='Marital_Status')


for i in A2.containers:
    A2.bar_label(i)

plt.title("Marital Status Wise Age Group Count")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

```python
Linechart = data.groupby("Sector",as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)

Linechart
```

```python
plt.figure(figsize=(15,6))

sns.lineplot(x="Sector",y="Amount",data=Linechart,marker="o")

plt.xticks(rotation=90)

plt.grid()

plt.title("Sector Wise Total Amount in Sorted Format")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()
```

```python
chart_2 = data.groupby("State",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
```

```python
plt.figure(figsize=(15,6))

sns.barplot(x="State",y="Orders",data=chart_2)

plt.xticks(rotation=90)

plt.title("Top 10 State Order Wise")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()

plt.show()
```

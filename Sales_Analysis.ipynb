# **Section 4 : Data Visualization**
In the exploratory analysis, I created multiple charts to understand customer demographics and sales distribution.
These visualizations revealed patterns in gender, age, marital status, zones, sectors, and states, helping identify key markets and revenue drivers.

# Chart 1: Gender-wise Count (Bar Chart)
Purpose: To analyze the distribution of customers by gender.

Observation: The bar chart shows how many male vs. female customers placed orders.

Insight: This helps identify whether sales are more concentrated among a particular gender group, useful for targeted marketing.

```python
plt.figure(figsize=(15,6))

A1 = sns.countplot(x="Zone",data=data,hue="Gender", edgecolor="black")

for i in A1.containers:
    A1.bar_label(i)

plt.title("Gender vs Zone Wise Count",color="#47116D",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

# Chart 2. Marital Status vs. Age Group (Bar Chart)
Purpose: To study customer demographics by combining marital status and age group.

Observation: The chart shows how different age groups are distributed across married and unmarried customers.

Insight: This reveals which demographic segments (e.g., young singles, middle-aged married) are most active in purchasing.

```python
plt.figure(figsize=(15,6))

A2 = sns.countplot(x='Age Group',data=data,hue='Marital_Status', edgecolor="black")


for i in A2.containers:
    A2.bar_label(i)

plt.title("Marital Status Wise Age Group Count",color="#47116D",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()
```

# Chart  3 : Sector-wise Amount (Line Chart)
Purpose: To track sales revenue across different sectors.

Observation: The line chart shows how the total amount varies by sector.

Insight: Identifies which sectors contribute the most to revenue, helping prioritize product focus.

```python
Linechart = data.groupby("Sector",as_index=False)["Amount"].sum().sort_values(by="Amount",ascending=False)

Linechart
```

```python
plt.figure(figsize=(15,6))

sns.lineplot(x="Sector",y="Amount",data=Linechart,marker="o")

plt.xticks(rotation=90)

plt.grid()

plt.title("Sector Wise Total Amount in Sorted Format",color="#47116D",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()
```

# Chart 4: Top 10 States by Orders (Bar Chart)
Purpose: To highlight the states with the highest number of orders.

Observation: The bar chart ranks states based on order volume.

Insight: Pinpoints the top-performing states, useful for resource allocation and regional campaigns.

```python
chart_2 = data.groupby("State",as_index=False)["Orders"].sum().sort_values(by="Orders",ascending=False).head(10)
```

```python
plt.figure(figsize=(15,6))

sns.barplot(x="State",y="Orders",data=chart_2, palette="tab20", edgecolor="black")

plt.xticks(rotation=90)

plt.title("Top 10 State Order Wise",color="#47116D",fontweight="bold", fontsize=16, family="sans-serif")

plt.xlabel("")

plt.ylabel("")

plt.show()

plt.show()

plt.show()
```

# Chart 5 : Zone-wise Amount Spent (Pie Chart)
Purpose: To visualize the proportion of total spending across zones.

Observation: The pie chart shows the percentage share of sales revenue by zone.

Insight: Provides a clear picture of revenue distribution, showing which zones dominate overall spending.

```python
pie_chart= data.groupby("Zone", as_index=False)["Amount"].sum()

plt.pie(pie_chart["Amount"], labels= pie_chart["Zone"], autopct = "%.f%%", startangle= 110)


plt.title("Zone Wise Amount", color="Purple", fontweight= "bold", family="georgia", fontsize= 16)
```
